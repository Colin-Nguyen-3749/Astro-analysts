```python
test_id = frame_ids[0]
response = requests.get(f'https://archive-api.lco.global/frames/{test_id}/headers/')
headers_data = response.json()
test_value = headers_data['data']['WCSERR']
print(test_value)
print(type(test_value))
```

```python
wcserrList = []
for frame in frame_ids:
    response = requests.get(f'https://archive-api.lco.global/frames/{frame}/headers/')
    headers_data = response.json()
    wcserrList.append(headers_data['data']['WCSERR'])
    
```

```python
wcserrList = []
problem_frames = []

for frame in frame_ids:
    response = requests.get(f'https://archive-api.lco.global/frames/{frame}/headers/')
    headers_data = response.json()
    if 'EXPOSE' == headers_data['data']['OBSTYPE']:
        wcserrList.append(headers_data['data']['WCSERR'])
    else:
        problem_frames.append(frame)

print(len(wcserrList))
print(problem_frames)
```
